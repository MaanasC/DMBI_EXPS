```python
# Select numerical attributes for analysis

numeric_cols = [
    "total_tickets",
    "price_per_ticket",
    "available_seats",
    "duration",
    "rating"
]

# Mean
print("===== MEAN =====")
print(eda_data[numeric_cols].mean())

# Median
print("\n===== MEDIAN =====")
print(eda_data[numeric_cols].median())

# Mode
print("\n===== MODE =====")
print(eda_data[numeric_cols].mode().iloc[0])

# Mode of important categorical attributes
categorical_cols = [
    "genre",
    "language",
    "city",
    "state"
]

print("\n===== CATEGORICAL MODES =====")

for col in categorical_cols:
    print(f"{col}: {eda_data[col].mode().iloc[0]}")
```

```python
# ===== STANDARD DEVIATION =====

print("===== STANDARD DEVIATION =====")
print(eda_data[numeric_cols].std())


# ===== VARIANCE =====

print("\n===== VARIANCE =====")
print(eda_data[numeric_cols].var())


# ===== RANGE =====

print("\n===== RANGE =====")

for col in numeric_cols:
    data_range = eda_data[col].max() - eda_data[col].min()
    print(f"{col}: {data_range}")


# ===== IQR =====

print("\n===== INTERQUARTILE RANGE (IQR) =====")

for col in numeric_cols:
    Q1 = eda_data[col].quantile(0.25)
    Q3 = eda_data[col].quantile(0.75)
    IQR = Q3 - Q1

    print(f"{col}: {IQR}")


# ===== POTENTIAL OUTLIERS =====

print("\n===== POTENTIAL OUTLIERS =====")

for col in numeric_cols:
    Q1 = eda_data[col].quantile(0.25)
    Q3 = eda_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = eda_data[
        (eda_data[col] < lower_bound) |
        (eda_data[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} potential outliers")
```

```python
# Calculate Pearson correlation

correlation_matrix = eda_data[numeric_cols].corr(method="pearson")

print("===== CORRELATION MATRIX =====")
display(correlation_matrix)


# Display strongest correlations

print("\n===== STRONGEST CORRELATIONS =====")

correlations = (
    correlation_matrix
    .unstack()
    .sort_values(ascending=False)
    .drop_duplicates()
)

display(correlations.head(10))
```
